## W6A1 Nhập vào một dòng gồm các số nguyên. Hãy loại bỏ phần tử trùng nhưng giữ nguyên thứ tự xuất hiện đầu tiên, rồi in ra list kết quả.
Yêu cầu in ra đúng dạng list Python, ví dụ [1, 2, 3]


```python
n = list(map(int,input("nhập 1 dãy số nguyên:").split()))
ds = []
for i in n:
    if i not in ds:
        ds.append(i)
print(ds)

```

## W6A2 Nhập một dòng số nguyên. In ra list tổng tích lũy (phần tử i là tổng từ đầu tới vị trí i), giữ nguyên thứ tự

## W6A4 Nhập một dòng gồm các cặp key:value cách nhau bởi dấu cách. Tạo dictionary ánh xạ mỗi key -> list các value theo thứ tự xuất hiện. Nếu key lặp lại, thêm vào list của key đó
 

```python
lst = list(map(str,input("nhập các cặp key:value:").split()))
dict = {}
for i in lst:
    key, value = i.split(":")
    value = int(value)
    if key not in dict:
        dict[key] = [value]
    else:
        dict[key].append(value)
        
print(dict)
```

## W6A13 [Đảo ngược key và value của dictionary]
Nhập vào một dictionary có value là duy nhất (không trùng nhau). Hãy tạo
một dictionary mới bằng cách đảo ngược key và value của dictionary ban
đầu. In ra dictionary mới.

```python
lst = list(map(str,input("nhập các cặp key:value:").split()))
dict = {}
for i in lst:
    key,val = i.split(":")
    val = int(val)
    dict[val] = key
print(dict)
```

## W6A11 [Tách số chẵn và lẻ từ tuple]
Nhập vào một tuple chứa các số nguyên. Hãy tạo ra hai tuple mới: một chứa
tất cả các số chẵn và một chứa tất cả các số lẻ từ tuple ban đầu, giữ nguyên
thứ tự tương đối. In ra hai tuple này trên hai dòng riêng biệt.


```python
lst = list(map(int,input("nhập 1 dãy số nguyên:").split()))
evtup = []
odtup = []
for i in lst:
    if i%2 == 0:
        evtup.append(i) 
    else:
        odtup.append(i)
print(tuple(evtup))
print(tuple(odtup))
```
