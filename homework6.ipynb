## W6A16 Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải


```python
# Bước 1: Nhập số hàng n và số cột m
n = int(input("Nhập số hàng n: "))
m = int(input("Nhập số cột m: "))

# Bước 2: Khởi tạo và nhập mảng hai chiều
matrix = []
print(f"Nhập các phần tử cho ma trận {n}x{m}:")
for i in range(n):
    row = []
    for j in range(m):
        val = int(input(f"Phần tử [{i}][{j}]: "))
        row.append(val)
    matrix.append(row)

# Bước 3: In ma trận với định dạng 4 vị trí căn phải
print("\nMa trận vừa nhập là:")
for row in matrix:
    for element in row:
        # :4 giúp dành ra 4 khoảng trống và căn phải
        print(f"{element:4}", end="") 
    print() # Xuống dòng sau khi in hết một hàng
```

## W6A17 Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng

```python
# Bước 1: Nhập số hàng và cột của ma trận vuông
n = int(input("Nhập số hàng và cột của ma trận vuông: "))

# Bước 2: Khởi tạo và nhập mảng hai chiều
matrix = []
print(f"Nhập các phần tử cho ma trận {n}x{n}:")
for i in range(n):
    row = []
    for j in range(n):
        val = int(input(f"Phần tử [{i}][{j}]: "))
        row.append(val)
    matrix.append(row)

# Bước 3: In ma trận với định dạng 4 vị trí căn phải
print("\nMa trận vừa nhập là:")
for row in matrix:
    for element in row:
        # :4 giúp dành ra 4 khoảng trống và căn phải
        print(f"{element:4}", end="") 
    print() # Xuống dòng sau khi in hết một hàng

# bước 4: in ra đường chéo chính
dcc = []
for i in range(n):
    dcc.append(matrix[i][i])
print(f"đường chéo chính của ma trận là: {dcc}")

# bước 5: in ra đường chéo phụ
dcp = []
for i in range(n):
    for k in range(n-1,1,-1):
        dcp.append(matrix[i][k])
print(f"đường chéo phụ của ma trận là: {dcp}")
```

## W6A20 Viết hàm đưa vào một list số nguyên và một số nguyên dương k. Hãy tìm và trả về vị trí của phần tử đầu tiên có giá trị k trong list, nếu không có thì trả về -1


```python
lst = list(map(int,input("nhập dãy số nguyên:").split()))
k = int(input("nhập 1 số nguyên: "))
for x in lst:
    if x == k:
        print(f"phần tử đầu tiên có giá trị bằng {k} là: {x}")
        break
else:
    print("-1")
```

## W6A21 Nhập một dòng gồm các chuỗi (tách bởi dấu cách). Tạo dictionary ánh xạ chuỗi -> tuple các vị trí (0-based) mà chuỗi đó xuất hiện


```python
lst = list(map(str,input("nhập chuỗi:").split()))
dict = {}
for x in range(len(lst)):
    if lst[x] not in dict:
        dict[lst[x]] = [x]
    else:
        dict[lst[x]].append(x)
print(lst)
print(dict)
```
