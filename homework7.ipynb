## W7A1 Viết hàm tìm kiếm nhị phân trên một danh sách đã được sắp xếp tăng dần. Trả về chỉ số của phần tử hoặc -1 nếu không tìm thấy. mẫu hàm: def binary_search(arr, left,right, target):

```python
def binary_search(lst, target):
    left, right = 0, len(lst) - 1
    while left <= right:
        mid = left + (right - left)//2
        if lst[mid] == target:
            return mid
        elif lst[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1

lst = list(map(int,input("nhập danh sách số nguyên đã sắp xếp: ").split()))
target = int(input("nhập số nguyên cần tìm: "))
print(binary_search(lst, target))
```

## W7A2 Cho danh sách arr và một số nguyên x, đếm xem x xuất hiện bao nhiêu lần trong danh sách. def count_occurrences(arr, x) -> int:


```python
def count_occurrences(arr, x):
    dem = 0
    if x not in arr:
        return 0
    else:
        for i in arr:
            if i == x:
                dem += 1
        return dem
arr = list(map(int, input("nhập danh sách số nguyên: ").split()))
x = int(input("nhập số nguyên cần đếm: "))
print(count_occurrences(arr, x))
```

## W7A16 Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu mỗi phần tử x trong mảng xuất hiện đúng x lần.
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp

```python
lst = list(map(int,input("nhập dãy số: ").split()))

def check_occurrences(lst):
    dict = {}
    for i in lst:
        if i not in dict:
            dict[i] = 1
        else:
            dict[i] += 1
    dem = 0
    for key,val in dict.items():
        if key == val:
            dem += 1
            continue
        else:
            if key < val:
                while key < val:
                    lst.remove(key)
                    val -= 1
            else:
                continue
    else:
        if dem == len(dict):
            return True
        else:
            return lst
        
print(check_occurrences(lst))
        
```

## W7A18 Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng dần, lẻ giảm dần

```python
def merge_sort_custom(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left = merge_sort_custom(arr[:mid])
    right = merge_sort_custom(arr[mid:])

    return merge_custom(left, right)

def is_better(a, b):
    """Trả về True nếu a nên đứng trước b theo đề bài"""
    # Trường hợp khác loại
    if a % 2 == 0 and b % 2 != 0: return True
    if a % 2 != 0 and b % 2 == 0: return False
    
    # Trường hợp cùng chẵn: tăng dần
    if a % 2 == 0 and b % 2 == 0:
        return a < b
    
    # Trường hợp cùng lẻ: giảm dần
    return a > b

def merge_custom(left, right):
    result = []
    i = j = 0

    while i < len(left) and j < len(right):
        if is_better(left[i], right[j]):
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])
    return result

# Chạy thử
data = list(map(int, input("Nhập dãy số: ").split()))
sorted_data = merge_sort_custom(data)
print(f"Kết quả: {sorted_data}")
```

## W7A20 Cho dãy N số nguyên dương, tính tổng các số có k chữ số. Dòng đầu ghi N số nguyên dương, dòng sau ghi số nguyên dương k

```python
lst = list(map(str,input("nhập dãy số: ").split()))
k = int(input("nhập k: "))
sumlst = []
for i in lst:
    if len(i) == k:
        tong = 0
        for j in i:
            tong += int(j)
        sumlst.append(tong)
        
print(sumlst)
```
